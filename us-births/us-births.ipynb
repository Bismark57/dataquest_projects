```python
def read_csv(UAE):
    UST = open(UAE).read()
    string_list = UST.split("\n")[1:]
    final_list = []
    
    for row in string_list:
        string_fields = row.split(",")
        int_fields = []
        for value in string_fields:
            int_fields.append(int(value))
            final_list.append(int_fields)
    return final_list

cdc_list = read_csv("US_births_1994-2003_CDC_NCHS.csv")


def month_births(list_of_list):
    births_per_month = dict()
    
    for i in list_of_list:
        month = i[1]
        births = i[4]
        if month in births_per_month:
            births_per_month[month] = births_per_month[month] + births
        else:
            births_per_month[month] = births
    return births_per_month

cdc_month_births = month_births(cdc_list)


def dow_births(list_of_list):
    day_of_week = {}
    
    for value in list_of_list:
        day = value[3]
        births = value[4]
        
        if day in day_of_week:
            day_of_week[day] = day_of_week[day] + births
        else:
            day_of_week[day] = births
    return day_of_week

cdc_day_births = dow_births(cdc_list)


## Combining all of the above into a single function

def calc_counts(data, column):
    called_value = {}
    
    for extracts in data:
        x = extracts[column]
        births = extracts[4]
        
        if x in called_value:
            called_value[x] = called_value[x] + births
        else:
            called_value[x] = births
    return called_value

cdc_year_births = calc_counts(cdc_list, 0)
cdc_month_births = calc_counts(cdc_list, 1)
cdc_dom_births = calc_counts(cdc_list, 2)
cdc_dow_births = calc_counts(cdc_list, 3)

cdc_year_births

        
        
    
    


    
```
