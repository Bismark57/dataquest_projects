```python
# We will base our analysis on atleast 5% of the brand names in the data
```

```python
selection = autos["brand"].value_counts(normalize=True)
brand_select = selection[selection > 0.05].index
brand_select
                                                            
```

```python
# The above brands falls within the greater than 5% values.
```

```python
aggregate_data = dict()

for brand in brand_select:
    match_brand = autos[autos["brand"] == brand]
    brand_price = match_brand["price_$"].mean()
    aggregate_data[brand] = int(brand_price)
brand_mean_prices = aggregate_data
brand_mean_prices
```

```python
# The data above indicates 'audi' as the manufacturer with the greatest mean and 'opel' with the least.
## This indicates how expensive audi cars can be with opel and ford being the least expensive and the rest fall inbetween.
```

```python
# Calculating the mean_mileage
```

```python
brand_mean_mileage = {}

for brand in brand_select:
    match_brand = autos[autos["brand"] == brand]
    brand_mileage = match_brand["odometer_km"].mean()
    brand_mean_mileage[brand] = int(brand_mileage)
brand_mean_mileage


mean_prices = pd.Series(brand_mean_prices).sort_values(ascending = False)
mean_mileage = pd.Series(brand_mean_mileage).sort_values(ascending = False)
```

```python
# Both prices and mileages need to be sorted out to align the data or get them working in the same dataframe
# panda methods pd.DataFrame and pd.Series are use to create both series objects and dataframe objects respectively
```

```python
bmp_series = pd.Series(mean_prices)
pd.DataFrame(bmp_series, columns=['mean_price'])
```

```python
bmm = pd.Series(mean_mileage)
pd.DataFrame(bmm, columns = ["mean_mileage"])
```

```python
# Grouping both mean_mileage and mean_price using the pandas 'pd.Dataframe' method to first assign to a variable and add the column to the frame
```

```python
new_df = pd.DataFrame(mean_mileage, columns = ["mean_mileage"])
new_df["mean_price"] = mean_prices
new_df
```
